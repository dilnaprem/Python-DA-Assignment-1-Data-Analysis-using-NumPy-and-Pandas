**Numpy Array Operations:**

**Problem Statement:**

 To learn how to use NumPy for creating arrays, performing basic
array operations, and conducting numerical computations through analyzing
temperature data.

**Scenario:**

You are given a list of daily average temperatures (in degrees Celsius)
recorded over two weeks. Your task is to use NumPy to analyze this data.

**Step wise Instructions:**

**1.Create a 1D NumPy Array:**

Create a 1D NumPy array named temperatures_w1 for Week 1 with the
following values:
[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

```python
import numpy as np
temperatures_w1=np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print(temperatures_w1)
```

**Imports the NumPy library.np.array()  Creates a NumPy array()
temperatures_w1 Stores the temperatures of seven days.
print() Displays the array.**

**2.Inspection and Properties:**

Inspect the shape, data type, and the number of elements in the array.
Inspect the shape of the temperatures_w1 array.

```python
print("shape:",temperatures_w1.shape)
print("data_type:",temperatures_w1.dtype)
print("number_of_elements:",temperatures_w1.size)
```

**shape shows the dimensions of the array.(7,) means the array is one-dimensional and contains seven elements. dtype shows the type of values stored,size shows how many elements are present.**


**3.Array Operations:**

• Convert the temperatures_w1 from Celsius to Fahrenheit using the
formula: Fahrenheit = (Celsius * 9/5) + 32

• Find the maximum, minimum, and mean temperatures for the week.

```python
Fahrenheit = (temperatures_w1 * 9/5) + 32
print(np.round(Fahrenheit, 2))
```

**The formula converts each temperature from Celsius to Fahrenheit. np.round() displays the values with two decimal places.**

```python
print("Maximum:", np.max(temperatures_w1))
print("Minimum:", np.min(temperatures_w1))
print("Mean:", np.round(np.mean(temperatures_w1), 2))
```

**np.max() finds the highest temperature, np.min() finds the lowest temperature, and np.mean() calculates the average temperature.**

**4.Array Slicing and Indexing:**

• Extract the temperatures for the first three days of the week.

• Extract the temperatures for the weekend (last two days).

• Extract the temperatures for the middle three days of the week.

```python
print(temperatures_w1[:3])
```

**[:3] selects the first three values.Here indexing starts from 0, and the ending index is not included.**

```python
print(temperatures_w1[-2:])
```

**[-2:] takes the last two values of the array**

```python
print(temperatures_w1[2:5])
```

**[2:5] selects the elements at index positions 2, 3, and 4.**

# Pandas Series:

1.Create a Pandas Series named marks with the following data:

➢ Marks: 95, 92, 89, 85, 80

➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd
marks=pd.Series([95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

**Pandas Series is created to store student marks.Each mark has a custom index label representing the students rank.**

**2.Indexing and Slicing:**

• Use the integer index position to access the mark of the 1st rank student.

• Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.

• Use the iloc accessor to retrieve the mark of the 3rd rank student.

• Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.

```python
print(marks.iloc[0])
```

**iloc[0] select the first element based on its position. Indexing starts from zero.**

```python
print(marks.loc['Rank1':'Rank3'])
```

**here loc select values using index labels.it retrieves the marks from Rank1 to Rank3.**

```python
print(marks.iloc[2])
```

**iloc[2] retrieve the third value because Python indexing starts at zero.**

```python
print(marks[marks > 90])
```

**Here the condition "marks>90" check each values and displays only the students who scored more than 90.**

# Manipulating Series:

• Modify the mark of the student with the 1st rank to 100.

• Remove the entry corresponding to the last rank from the Series.

• Compute the CGPA by dividing each mark in the Series by 10.

```python
marks.loc['Rank1'] = 100
print(marks)
```

**loc is used to select Rank1 and update its mark from 95 to 100.**

```python
marks = marks.drop('Rank5')
print(marks)
```

**drop() removes the selected index label**

```python
cgpa = marks / 10
print(cgpa)
```

**To calculate the CGPA each student's mark is divided by 10**

**1.Creating Pandas DataFrame:**

• Create a Pandas DataFrame named transactions with the following data:

➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110

➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'

**3. Manipulating the DataFrame:**

a) Modify the 'Amount' for the transaction with TransactionID 102 to 165.

b) Add a new column 'Discount' by calculating 10% of the 'Amount'.

c) Remove the row with TransactionID 109.

d) Delete the 'Discount' column from the DataFrame.

a) Modify the 'Amount' for the transaction with TransactionID 102 to 165.


```python
transactions.loc[transactions['TransactionID']==102,'Amount']=165
display(transactions)
```

**loc is used to find TransactionID 102 and change its Amount from 150 to 165. The remaining rows stay unchanged.**

b) Add a new column 'Discount' by calculating 10% of the 'Amount'.


```python
transactions['Discount']=transactions['Amount']*0.10
display(transactions)
```

**A new column named Discount is created by multiplying each transaction amount by 0.10, which calculates a 10% discount.**

c) Remove the row with TransactionID 109.


```python
transactions=transactions[transactions['TransactionID']!=109]
display(transactions)
```

**Here !=109 means not equal to 109.The code keeps all rows except TransactionID 109 and saves the result back into transactions.**

d) Delete the 'Discount' column from the DataFrame.

```python
transactions=transactions.drop(columns=['Discount'])
display(transactions)
```

**drop(columns=['Discount']) remove the Discount column.The updated DataFrame is saved back into transactions.**
